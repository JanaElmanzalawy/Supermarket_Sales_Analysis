# Data Analysis

```python
df.columns
```

```python
# Total Revenue
Total_Revenue = df['Sales'].sum().item()
print(f"Total Revenue is {Total_Revenue} ")
```

```python
# Revenue by City
Revenue_by_City = df.groupby('City')['Sales'].sum().sort_values(ascending=False)
Revenue_by_City
```

```python
highest_Revenue_by_City= df.groupby('City')['Sales'].sum().sort_values(ascending=False).head(1)
highest_Revenue_by_City
```

```python
plt.figure(figsize=(8,5))

sns.barplot(x=Revenue_by_City.index, y=Revenue_by_City.values,palette='Set2')

plt.title('Total Revenue by City')
plt.xlabel('City')
plt.ylabel('Total Revenue')
plt.tight_layout()
plt.show()
```

```python
# Profit by Branch
profit_by_branch = df.groupby('Branch')['Gross_Income'].sum().sort_values(ascending=False)
profit_by_branch
```

```python
plt.figure(figsize=(8,5))

sns.barplot(x=profit_by_branch.index, y=profit_by_branch.values,palette='Set2')

plt.title('Profit by Branch')
plt.xlabel('Branch')
plt.ylabel('Profit')
plt.tight_layout()
plt.show()
```

```python
#  Revenue and Profit by Product Category
Revenue_Profit_by_Product_Category = df.groupby('Product_Line').agg(Revenue=('Sales', 'sum'),Profit=('Gross_Income', 'sum')).sort_values('Revenue', ascending=False)
Revenue_Profit_by_Product_Category
```

```python
top_category = df.groupby('Product_Line').agg(Revenue=('Sales', 'sum'),Profit=('Gross_Income', 'sum')).sort_values('Revenue', ascending=False).head(1)
top_category
```

```python
plt.figure(figsize=(12,6))

sns.barplot(data=Revenue_Profit_by_Product_Category.reset_index().melt(id_vars='Product_Line',value_vars=['Revenue', 'Profit'],var_name='Type',value_name='Amount'),
    x='Product_Line',
    y='Amount',
    hue='Type',
    palette='Set2'
)

plt.title('Revenue and Profit by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Amount')
plt.xticks(rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
# Spending by Customer Type
Spending_by_Customer = df.groupby('Customer_Type')['Sales'].sum().sort_values(ascending= False)
Spending_by_Customer
```

```python
plt.figure(figsize=(8,5))

sns.barplot(x=Spending_by_Customer.index, y=Spending_by_Customer.values,palette='Set2')

plt.title('Spending by Customer Typ')
plt.xlabel('Customer Type')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()
```

```python
spend_more = df.groupby('Customer_Type')['Sales'].sum().sort_values(ascending= False).head(1)
spend_more
```

```python
# Payment Method
payment_method = df['Payment'].value_counts().sort_values(ascending=False)
payment_method
```

```python
top_payment_method = df['Payment'].value_counts().sort_values(ascending=False).head(1)
top_payment_method
```

```python
# Customer Satisfaction by Branch
satisfaction_by_branch = (df.groupby('Branch')['Satisfied'].apply(lambda x: (x == 'Satisfied').mean() * 100).sort_values(ascending=False))
satisfaction_by_branch
```
